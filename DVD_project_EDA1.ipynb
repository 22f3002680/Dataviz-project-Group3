## Customers dataset

### Data source

This notebook originally read CSVs from Google Colab (`/content`, Google Drive). It now loads the **same tables from the project workbook** `Dataviz_proj_all_datasets.xlsx`, which is **downloaded separately** (no longer stored in git). Place the workbook at the repository root, or set the `DVD_WORKBOOK` environment variable to its path, then run all cells.

```python
# --- Portable data loader (replaces the original Colab /content paths) ---
import os
from pathlib import Path
import pandas as pd

_ROOT = Path.cwd()
while _ROOT != _ROOT.parent and not (_ROOT / "Dataviz_proj_all_datasets.xlsx").exists():
    _ROOT = _ROOT.parent
WORKBOOK = Path(os.environ.get("DVD_WORKBOOK", _ROOT / "Dataviz_proj_all_datasets.xlsx"))

# CSV base name -> workbook sheet name (some sheet names are truncated in the workbook)
_SHEET = {
    "customers_dataset": "customers_dataset",
    "orders_dataset": "orders_dataset",
    "order_items_dataset": "order_items_dataset",
    "order_reviews_dataset": "order_reviews_dataset",
    "order_payments_dataset": "order_payments_dataset",
    "products_dataset": "products_dataset",
    "sellers_dataset": "sellers_dataset",
    "geolocation_dataset": "geolocation_dataset",
    "marketing_qualified_leads_dataset": "marketing_qualified_leads_datas",
    "closed_deals_dataset": "closed_deals",
    "product_category_name_translation": "product_category_name_translati",
}
_BOOK = pd.ExcelFile(WORKBOOK)

def load(name):
    """Load a dataset by its CSV base name from the project workbook."""
    key = os.path.basename(str(name)).replace(".csv", "")
    return pd.read_excel(_BOOK, sheet_name=_SHEET.get(key, key))
```

```python
print("\n========== DATASET OVERVIEW ==========")

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()


# ==========================================================
# 3. MISSING VALUES
# ==========================================================

print("\n========== MISSING VALUES ==========")

print("Missing values in each column:")
print(df.isnull().sum())

print("\nMissing values (%):")
print(
    (df.isnull().sum() / len(df) * 100)
    .round(2)
)


# ==========================================================
# 4. DUPLICATE RECORDS
# ==========================================================

print("\n========== DUPLICATES ==========")

print("Number of duplicate rows:")
print(df.duplicated().sum())


# ==========================================================
# 5. UNIQUE VALUES
# ==========================================================

print("\n========== UNIQUE VALUES ==========")

print(df.nunique().sort_values(ascending=False))


# ==========================================================
# 6. NUMERICAL VARIABLES
# ==========================================================

print("\n========== NUMERICAL VARIABLES ==========")

numerical_cols = df.select_dtypes(
    include=np.number
).columns

print("Numerical columns:")
print(numerical_cols.tolist())

print("\nDescriptive statistics:")
print(df[numerical_cols].describe().T)


# ==========================================================
# 7. CATEGORICAL VARIABLES
# ==========================================================

print("\n========== CATEGORICAL VARIABLES ==========")

categorical_cols = df.select_dtypes(
    include="object"
).columns

print("Categorical columns:")
print(categorical_cols.tolist())

for col in categorical_cols:
    print(f"\n--- {col} ---")
    print("Number of unique values:", df[col].nunique())
    print("\nValue counts:")
    print(df[col].value_counts().head(20))


# ==========================================================
# 8. PAYMENT TYPE ANALYSIS
# ==========================================================

print("\n========== PAYMENT TYPE ==========")

print(
    df["payment_type"].value_counts()
)

print("\nPayment type percentage:")
print(
    df["payment_type"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)


# ==========================================================
# 9. PAYMENT INSTALLMENTS
# ==========================================================

print("\n========== PAYMENT INSTALLMENTS ==========")

print(
    df["payment_installments"]
    .value_counts()
    .sort_index()
)

print("\nInstallment statistics:")
print(
    df["payment_installments"].describe()
)


# ==========================================================
# 10. PAYMENT VALUE
# ==========================================================

print("\n========== PAYMENT VALUE ==========")

print("Minimum payment:",
      df["payment_value"].min())

print("Maximum payment:",
      df["payment_value"].max())

print("Mean payment:",
      df["payment_value"].mean())

print("Median payment:",
      df["payment_value"].median())

print("\nPayment value statistics:")
print(
    df["payment_value"].describe()
)


# ==========================================================
# 11. CHECK FOR SUSPICIOUS VALUES
# ==========================================================

print("\n========== DATA QUALITY CHECKS ==========")

print("Payment values <= 0:")
print(
    (df["payment_value"] <= 0).sum()
)

print("Installments <= 0:")
print(
    (df["payment_installments"] <= 0).sum()
)

print("Payment sequential <= 0:")
print(
    (df["payment_sequential"] <= 0).sum()
)


# ==========================================================
# 12. PAYMENT SEQUENCE
# ==========================================================

print("\n========== PAYMENT SEQUENCE ==========")

print(
    df["payment_sequential"]
    .value_counts()
    .sort_index()
)

print("\nPayment sequence statistics:")
print(
    df["payment_sequential"].describe()
)


# ==========================================================
# 13. ORDERS WITH MULTIPLE PAYMENTS
# ==========================================================

print("\n========== MULTIPLE PAYMENTS PER ORDER ==========")

payment_counts = df["order_id"].value_counts()

print("Unique orders:",
      df["order_id"].nunique())

print("Orders with more than one payment:",
      (payment_counts > 1).sum())

print("\nNumber of payments per order:")
print(payment_counts.describe())


# ==========================================================
# 14. CHECK PAYMENT SEQUENTIAL CONSISTENCY
# ==========================================================

print("\n========== PAYMENT SEQUENCE CHECK ==========")

max_sequence = (
    df.groupby("order_id")["payment_sequential"]
    .max()
)

print("Maximum payment sequence:",
      max_sequence.max())

print("\nOrders with multiple payment sequences:")
print(
    max_sequence[max_sequence > 1].head(20)
)


# ==========================================================
# 15. PAYMENT VALUE BY PAYMENT TYPE
# ==========================================================

print("\n========== PAYMENT VALUE BY TYPE ==========")

payment_summary = (
    df.groupby("payment_type")["payment_value"]
    .agg([
        "count",
        "sum",
        "mean",
        "median",
        "min",
        "max"
    ])
    .sort_values("sum", ascending=False)
)

print(payment_summary)


# ==========================================================
# 16. INSTALLMENTS BY PAYMENT TYPE
# ==========================================================

print("\n========== INSTALLMENTS BY PAYMENT TYPE ==========")

installment_summary = (
    df.groupby("payment_type")["payment_installments"]
    .agg([
        "count",
        "mean",
        "median",
        "min",
        "max"
    ])
)

print(installment_summary)


# ==========================================================
# 17. OUTLIER CHECK USING IQR
# ==========================================================

print("\n========== OUTLIER CHECK ==========")

Q1 = df["payment_value"].quantile(0.25)
Q3 = df["payment_value"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["payment_value"] < lower_bound) |
    (df["payment_value"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

print("Number of potential payment-value outliers:",
      len(outliers))


# ==========================================================
# 18. FINAL SUMMARY
# ==========================================================

print("\n========== FINAL SUMMARY ==========")

print("Rows:", len(df))
print("Columns:", len(df.columns))
print("Unique orders:", df["order_id"].nunique())
print("Unique payment types:", df["payment_type"].nunique())
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

## Order reviews dataset

```python
# ==========================================================
# 1. LOAD DATASET
# ==========================================================

df = load("order_reviews_dataset")

print("Dataset loaded successfully.")


# ==========================================================
# 2. BASIC DATASET INFORMATION
# ==========================================================

print("\n========== DATASET OVERVIEW ==========")

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
print(df.head())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()


# ==========================================================
# 3. MISSING VALUES
# ==========================================================

print("\n========== MISSING VALUES ==========")

missing = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

print(missing)


# ==========================================================
# 4. DUPLICATES
# ==========================================================

print("\n========== DUPLICATES ==========")

print("Duplicate rows:", df.duplicated().sum())

print("Duplicate review IDs:",
      df["review_id"].duplicated().sum())


# ==========================================================
# 5. UNIQUE VALUES
# ==========================================================

print("\n========== UNIQUE VALUES ==========")

print(df.nunique().sort_values(ascending=False))


# ==========================================================
# 6. REVIEW SCORE
# ==========================================================

print("\n========== REVIEW SCORE ==========")

print("Review score distribution:")
print(df["review_score"].value_counts().sort_index())

print("\nReview score percentage:")
print(
    df["review_score"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)

print("\nReview score statistics:")
print(df["review_score"].describe())


# ==========================================================
# 7. CHECK VALID REVIEW SCORES
# ==========================================================

print("\n========== REVIEW SCORE VALIDATION ==========")

print("Scores below 1:",
      (df["review_score"] < 1).sum())

print("Scores above 5:",
      (df["review_score"] > 5).sum())

print("Unique review scores:",
      sorted(df["review_score"].unique()))


# ==========================================================
# 8. REVIEW COMMENTS
# ==========================================================

print("\n========== REVIEW COMMENTS ==========")

print("Reviews with a title:",
      df["review_comment_title"].notna().sum())

print("Reviews without a title:",
      df["review_comment_title"].isna().sum())

print("Reviews with a message:",
      df["review_comment_message"].notna().sum())

print("Reviews without a message:",
      df["review_comment_message"].isna().sum())


# ==========================================================
# 9. COMMENT LENGTH
# ==========================================================

print("\n========== COMMENT LENGTH ==========")

# Replace missing comments with empty strings
df["comment_length"] = (
    df["review_comment_message"]
    .fillna("")
    .str.len()
)

print(df["comment_length"].describe())

print("\nReviews with no written message:",
      (df["comment_length"] == 0).sum())

print("\nReviews with written message:",
      (df["comment_length"] > 0).sum())


# ==========================================================
# 10. TITLE LENGTH
# ==========================================================

df["title_length"] = (
    df["review_comment_title"]
    .fillna("")
    .str.len()
)

print("\n========== REVIEW TITLE LENGTH ==========")

print(df["title_length"].describe())

print("Reviews with no title:",
      (df["title_length"] == 0).sum())


# ==========================================================
# 11. DATE CONVERSION
# ==========================================================

print("\n========== DATE CONVERSION ==========")

df["review_creation_date"] = pd.to_datetime(
    df["review_creation_date"],
    errors="coerce"
)

df["review_answer_timestamp"] = pd.to_datetime(
    df["review_answer_timestamp"],
    errors="coerce"
)

print("Review creation date type:",
      df["review_creation_date"].dtype)

print("Review answer timestamp type:",
      df["review_answer_timestamp"].dtype)

print("\nInvalid creation dates:",
      df["review_creation_date"].isna().sum())

print("Invalid answer timestamps:",
      df["review_answer_timestamp"].isna().sum())


# ==========================================================
# 12. DATE RANGE
# ==========================================================

print("\n========== DATE RANGE ==========")

print("Earliest review:",
      df["review_creation_date"].min())

print("Latest review:",
      df["review_creation_date"].max())

print("Earliest answer:",
      df["review_answer_timestamp"].min())

print("Latest answer:",
      df["review_answer_timestamp"].max())


# ==========================================================
# 13. REVIEW RESPONSE TIME
# ==========================================================

print("\n========== REVIEW RESPONSE TIME ==========")

df["response_time"] = (
    df["review_answer_timestamp"]
    - df["review_creation_date"]
)

df["response_hours"] = (
    df["response_time"].dt.total_seconds() / 3600
)

print(df["response_hours"].describe())

print("\nReviews answered before creation:",
      (df["response_hours"] < 0).sum())


# ==========================================================
# 14. ORDERS AND REVIEWS
# ==========================================================

print("\n========== ORDERS AND REVIEWS ==========")

print("Unique orders:",
      df["order_id"].nunique())

print("Unique reviews:",
      df["review_id"].nunique())

reviews_per_order = df["order_id"].value_counts()

print("\nReviews per order:")
print(reviews_per_order.describe())

print("\nOrders with more than one review:",
      (reviews_per_order > 1).sum())


# ==========================================================
# 15. REVIEW SCORE BY WRITTEN COMMENT
# ==========================================================

print("\n========== SCORE VS WRITTEN REVIEW ==========")

df["has_comment"] = (
    df["review_comment_message"].notna()
)

print(
    df.groupby("has_comment")["review_score"]
    .agg(["count", "mean", "median"])
)


# ==========================================================
# 16. REVIEW SCORE BY REVIEW TITLE
# ==========================================================

df["has_title"] = (
    df["review_comment_title"].notna()
)

print("\n========== SCORE VS REVIEW TITLE ==========")

print(
    df.groupby("has_title")["review_score"]
    .agg(["count", "mean", "median"])
)


# ==========================================================
# 17. REVIEW SCORE GROUPS
# ==========================================================

print("\n========== SCORE GROUPS ==========")

df["review_category"] = pd.cut(
    df["review_score"],
    bins=[0, 2, 3, 5],
    labels=[
        "Negative (1-2)",
        "Neutral (3)",
        "Positive (4-5)"
    ]
)

print(df["review_category"].value_counts())

print("\nPercentage:")
print(
    df["review_category"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)


# ==========================================================
# 18. FINAL DATA QUALITY CHECK
# ==========================================================

print("\n========== FINAL DATA QUALITY CHECK ==========")

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("Missing values:",
      df.isnull().sum().sum())

print("Duplicate rows:",
      df.duplicated().sum())

print("Unique orders:",
      df["order_id"].nunique())

print("Unique reviews:",
      df["review_id"].nunique())

print("Average review score:",
      df["review_score"].mean())

print("Median review score:",
      df["review_score"].median())
```

## Orders dataset

```python
import pandas as pd
import numpy as np

# ==========================================================
# 1. LOAD DATASET
# ==========================================================

df = load("orders_dataset")

print("Dataset loaded successfully.")


# ==========================================================
# 2. BASIC DATASET INFORMATION
# ==========================================================

print("\n========== DATASET OVERVIEW ==========")

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
print(df.head())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()


# ==========================================================
# 3. MISSING VALUES
# ==========================================================

print("\n========== MISSING VALUES ==========")

missing = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

print(missing)


# ==========================================================
# 4. DUPLICATES
# ==========================================================

print("\n========== DUPLICATES ==========")

print("Duplicate rows:", df.duplicated().sum())

print("Duplicate order IDs:",
      df["order_id"].duplicated().sum())

print("Duplicate customer IDs:",
      df["customer_id"].duplicated().sum())


# ==========================================================
# 5. UNIQUE VALUES
# ==========================================================

print("\n========== UNIQUE VALUES ==========")

print(df.nunique().sort_values(ascending=False))


# ==========================================================
# 6. ORDER STATUS
# ==========================================================

print("\n========== ORDER STATUS ==========")

print("Order status counts:")
print(df["order_status"].value_counts())

print("\nOrder status percentages:")
print(
    df["order_status"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)


# ==========================================================
# 7. CHECK ALL ORDER STATUSES
# ==========================================================

print("\nUnique order statuses:")
print(df["order_status"].unique())

print("\nNumber of order statuses:",
      df["order_status"].nunique())


# ==========================================================
# 8. CONVERT DATE COLUMNS
# ==========================================================

print("\n========== DATE CONVERSION ==========")

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    df[col] = pd.to_datetime(
        df[col],
        errors="coerce"
    )

print("\nUpdated data types:")
print(df[date_columns].dtypes)


# ==========================================================
# 9. CHECK INVALID DATES
# ==========================================================

print("\n========== INVALID DATE VALUES ==========")

for col in date_columns:
    print(
        col,
        "->",
        df[col].isna().sum(),
        "missing/invalid values"
    )


# ==========================================================
# 10. DATE RANGES
# ==========================================================

print("\n========== DATE RANGES ==========")

for col in date_columns:
    print(
        f"{col}:",
        df[col].min(),
        "to",
        df[col].max()
    )


# ==========================================================
# 11. ORDER PURCHASE DATE ANALYSIS
# ==========================================================

print("\n========== PURCHASE DATE ==========")

print("Earliest order:",
      df["order_purchase_timestamp"].min())

print("Latest order:",
      df["order_purchase_timestamp"].max())


# ==========================================================
# 12. EXTRACT DATE COMPONENTS
# ==========================================================

df["purchase_year"] = (
    df["order_purchase_timestamp"].dt.year
)

df["purchase_month"] = (
    df["order_purchase_timestamp"].dt.month
)

df["purchase_day"] = (
    df["order_purchase_timestamp"].dt.day
)

df["purchase_day_of_week"] = (
    df["order_purchase_timestamp"].dt.day_name()
)

df["purchase_hour"] = (
    df["order_purchase_timestamp"].dt.hour
)

print("\n========== PURCHASE DATE COMPONENTS ==========")

print("Years:")
print(df["purchase_year"].value_counts().sort_index())

print("\nMonths:")
print(df["purchase_month"].value_counts().sort_index())

print("\nDays of week:")
print(df["purchase_day_of_week"].value_counts())

print("\nPurchase hours:")
print(df["purchase_hour"].value_counts().sort_index())


# ==========================================================
# 13. ORDER PROCESSING TIME
# ==========================================================

print("\n========== ORDER PROCESSING TIME ==========")

df["approval_time_hours"] = (
    df["order_approved_at"]
    - df["order_purchase_timestamp"]
).dt.total_seconds() / 3600

print("Approval time statistics:")
print(df["approval_time_hours"].describe())


# ==========================================================
# 14. DELIVERY TO CARRIER TIME
# ==========================================================

df["carrier_processing_hours"] = (
    df["order_delivered_carrier_date"]
    - df["order_approved_at"]
).dt.total_seconds() / 3600

print("\n========== CARRIER PROCESSING TIME ==========")

print(df["carrier_processing_hours"].describe())


# ==========================================================
# 15. DELIVERY TIME
# ==========================================================

df["delivery_time_days"] = (
    df["order_delivered_customer_date"]
    - df["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 3600)

print("\n========== DELIVERY TIME ==========")

print(df["delivery_time_days"].describe())


# ==========================================================
# 16. ESTIMATED DELIVERY TIME
# ==========================================================

df["estimated_delivery_days"] = (
    df["order_estimated_delivery_date"]
    - df["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 3600)

print("\n========== ESTIMATED DELIVERY TIME ==========")

print(df["estimated_delivery_days"].describe())


# ==========================================================
# 17. ACTUAL VS ESTIMATED DELIVERY
# ==========================================================

df["delivery_difference_days"] = (
    df["order_delivered_customer_date"]
    - df["order_estimated_delivery_date"]
).dt.total_seconds() / (24 * 3600)

print("\n========== ACTUAL VS ESTIMATED DELIVERY ==========")

print(df["delivery_difference_days"].describe())


# ==========================================================
# 18. DELIVERY PERFORMANCE
# ==========================================================

print("\n========== DELIVERY PERFORMANCE ==========")

delivered = df[
    df["order_delivered_customer_date"].notna()
].copy()

delivered["delivery_status"] = np.where(
    delivered["delivery_difference_days"] <= 0,
    "On or before estimated date",
    "Delivered late"
)

print(
    delivered["delivery_status"].value_counts()
)

print("\nDelivery performance percentage:")
print(
    delivered["delivery_status"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)


# ==========================================================
# 19. CHECK FOR NEGATIVE DURATIONS
# ==========================================================

print("\n========== DATA CONSISTENCY CHECK ==========")

print(
    "Negative approval time:",
    (df["approval_time_hours"] < 0).sum()
)

print(
    "Negative carrier processing time:",
    (df["carrier_processing_hours"] < 0).sum()
)

print(
    "Negative delivery time:",
    (df["delivery_time_days"] < 0).sum()
)


# ==========================================================
# 20. ORDER STATUS VS MISSING DELIVERY DATE
# ==========================================================

print("\n========== STATUS VS DELIVERY DATE ==========")

print(
    pd.crosstab(
        df["order_status"],
        df["order_delivered_customer_date"].isna(),
        margins=True
    )
)


# ==========================================================
# 21. CUSTOMER ORDER COUNTS
# ==========================================================

print("\n========== ORDERS PER CUSTOMER ==========")

orders_per_customer = (
    df["customer_id"]
    .value_counts()
)

print(orders_per_customer.describe())

print("\nCustomers with more than one order:",
      (orders_per_customer > 1).sum())


# ==========================================================
# 22. FINAL DATA QUALITY SUMMARY
# ==========================================================

print("\n========== FINAL SUMMARY ==========")

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("Unique orders:",
      df["order_id"].nunique())

print("Unique customers:",
      df["customer_id"].nunique())

print("Unique order statuses:",
      df["order_status"].nunique())

print("Duplicate rows:",
      df.duplicated().sum())

print("Total missing values:",
      df.isnull().sum().sum())
```

```python
import pandas as pd
import numpy as np

# ==========================================================
# 1. LOAD DATASET
# ==========================================================

df = load("closed_deals_dataset")

print("Dataset loaded successfully.")


# ==========================================================
# 2. BASIC DATASET INFORMATION
# ==========================================================

print("\n========== DATASET OVERVIEW ==========")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
print(df.head())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()


# ==========================================================
# 3. MISSING VALUES
# ==========================================================

print("\n========== MISSING VALUES ==========")

missing = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage":
        (df.isnull().sum() / len(df) * 100).round(2)
})

print(missing.sort_values(
    "Missing_Percentage",
    ascending=False
))


# ==========================================================
# 4. DUPLICATES
# ==========================================================

print("\n========== DUPLICATES ==========")

print("Duplicate rows:", df.duplicated().sum())

print("Duplicate MQL IDs:",
      df["mql_id"].duplicated().sum())

print("Duplicate seller IDs:",
      df["seller_id"].duplicated().sum())


# ==========================================================
# 5. UNIQUE VALUES
# ==========================================================

print("\n========== UNIQUE VALUES ==========")

print(df.nunique().sort_values(ascending=False))


# ==========================================================
# 6. CATEGORICAL COLUMNS
# ==========================================================

print("\n========== CATEGORICAL VARIABLES ==========")

categorical_cols = df.select_dtypes(
    include="object"
).columns

print("Categorical columns:")
print(categorical_cols.tolist())

for col in categorical_cols:

    print(f"\n--- {col} ---")

    print("Unique values:",
          df[col].nunique())

    print("\nValue counts:")
    print(
        df[col]
        .value_counts(dropna=False)
        .head(20)
    )


# ==========================================================
# 7. BUSINESS SEGMENT
# ==========================================================

print("\n========== BUSINESS SEGMENT ==========")

print(
    df["business_segment"]
    .value_counts(dropna=False)
)

print("\nPercentage:")
print(
    df["business_segment"]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
)


# ==========================================================
# 8. LEAD TYPE
# ==========================================================

print("\n========== LEAD TYPE ==========")

print(
    df["lead_type"]
    .value_counts(dropna=False)
)

print("\nPercentage:")
print(
    df["lead_type"]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
)


# ==========================================================
# 9. LEAD BEHAVIOUR PROFILE
# ==========================================================

print("\n========== LEAD BEHAVIOUR PROFILE ==========")

print(
    df["lead_behaviour_profile"]
    .value_counts(dropna=False)
)


# ==========================================================
# 10. COMPANY / GTIN
# ==========================================================

print("\n========== COMPANY / GTIN ==========")

print("Has company:")
print(
    df["has_company"]
    .value_counts(dropna=False)
)

print("\nHas GTIN:")
print(
    df["has_gtin"]
    .value_counts(dropna=False)
)


# ==========================================================
# 11. BUSINESS TYPE
# ==========================================================

print("\n========== BUSINESS TYPE ==========")

print(
    df["business_type"]
    .value_counts(dropna=False)
)


# ==========================================================
# 12. AVERAGE STOCK
# ==========================================================

print("\n========== AVERAGE STOCK ==========")

print(
    df["average_stock"]
    .value_counts(dropna=False)
)

print("\nData type:")
print(df["average_stock"].dtype)


# ==========================================================
# 13. CONVERT AVERAGE STOCK TO NUMERIC
# ==========================================================

df["average_stock_numeric"] = pd.to_numeric(
    df["average_stock"],
    errors="coerce"
)

print("\nNumeric average stock statistics:")
print(
    df["average_stock_numeric"].describe()
)


# ==========================================================
# 14. NUMERICAL VARIABLES
# ==========================================================

print("\n========== NUMERICAL VARIABLES ==========")

numerical_cols = df.select_dtypes(
    include=np.number
).columns

print("Numerical columns:")
print(numerical_cols.tolist())

print("\nDescriptive statistics:")
print(
    df[numerical_cols]
    .describe()
    .T
)


# ==========================================================
# 15. PRODUCT CATALOG SIZE
# ==========================================================

print("\n========== PRODUCT CATALOG SIZE ==========")

print(
    df["declared_product_catalog_size"]
    .describe()
)

print("Zero catalog size:",
      (df["declared_product_catalog_size"] == 0).sum())

print("Negative catalog size:",
      (df["declared_product_catalog_size"] < 0).sum())


# ==========================================================
# 16. MONTHLY REVENUE
# ==========================================================

print("\n========== DECLARED MONTHLY REVENUE ==========")

print(
    df["declared_monthly_revenue"]
    .describe()
)

print("Zero revenue:",
      (df["declared_monthly_revenue"] == 0).sum())

print("Negative revenue:",
      (df["declared_monthly_revenue"] < 0).sum())


# ==========================================================
# 17. WON DATE
# ==========================================================

print("\n========== WON DATE ==========")

df["won_date"] = pd.to_datetime(
    df["won_date"],
    errors="coerce"
)

print("Data type:")
print(df["won_date"].dtype)

print("\nInvalid dates:")
print(df["won_date"].isna().sum())

print("\nEarliest closed deal:")
print(df["won_date"].min())

print("\nLatest closed deal:")
print(df["won_date"].max())


# ==========================================================
# 18. DATE COMPONENTS
# ==========================================================

df["won_year"] = df["won_date"].dt.year
df["won_month"] = df["won_date"].dt.month
df["won_day"] = df["won_date"].dt.day
df["won_day_of_week"] = df["won_date"].dt.day_name()
df["won_hour"] = df["won_date"].dt.hour

print("\n========== DATE COMPONENTS ==========")

print("Deals by year:")
print(
    df["won_year"]
    .value_counts()
    .sort_index()
)

print("\nDeals by month:")
print(
    df["won_month"]
    .value_counts()
    .sort_index()
)

print("\nDeals by day of week:")
print(
    df["won_day_of_week"]
    .value_counts()
)

print("\nDeals by hour:")
print(
    df["won_hour"]
    .value_counts()
    .sort_index()
)


# ==========================================================
# 19. SELLER ANALYSIS
# ==========================================================

print("\n========== SELLER ANALYSIS ==========")

print("Unique sellers:",
      df["seller_id"].nunique())

seller_deals = (
    df["seller_id"]
    .value_counts()
)

print("\nDeals per seller:")
print(
    seller_deals.describe()
)

print("\nTop 20 sellers by closed deals:")
print(
    seller_deals.head(20)
)


# ==========================================================
# 20. SDR ANALYSIS
# ==========================================================

print("\n========== SDR ANALYSIS ==========")

print("Unique SDRs:",
      df["sdr_id"].nunique())

sdr_deals = (
    df["sdr_id"]
    .value_counts()
)

print("\nDeals per SDR:")
print(
    sdr_deals.describe()
)

print("\nTop SDRs:")
print(
    sdr_deals.head(20)
)


# ==========================================================
# 21. SALES REPRESENTATIVE ANALYSIS
# ==========================================================

print("\n========== SALES REPRESENTATIVE ==========")

print("Unique SRs:",
      df["sr_id"].nunique())

sr_deals = (
    df["sr_id"]
    .value_counts()
)

print("\nDeals per SR:")
print(
    sr_deals.describe()
)

print("\nTop SRs:")
print(
    sr_deals.head(20)
)


# ==========================================================
# 22. REVENUE BY BUSINESS SEGMENT
# ==========================================================

print("\n========== BUSINESS SEGMENT SUMMARY ==========")

segment_summary = (
    df.groupby("business_segment", dropna=False)
    .agg(
        deals=("mql_id", "count"),
        avg_revenue=("declared_monthly_revenue", "mean"),
        median_revenue=("declared_monthly_revenue", "median"),
        total_revenue=("declared_monthly_revenue", "sum"),
        avg_catalog_size=(
            "declared_product_catalog_size",
            "mean"
        )
    )
    .sort_values(
        "deals",
        ascending=False
    )
)

print(segment_summary)


# ==========================================================
# 23. REVENUE BY BUSINESS TYPE
# ==========================================================

print("\n========== BUSINESS TYPE SUMMARY ==========")

business_summary = (
    df.groupby("business_type", dropna=False)
    .agg(
        deals=("mql_id", "count"),
        avg_revenue=("declared_monthly_revenue", "mean"),
        median_revenue=("declared_monthly_revenue", "median"),
        total_revenue=("declared_monthly_revenue", "sum")
    )
    .sort_values(
        "deals",
        ascending=False
    )
)

print(business_summary)


# ==========================================================
# 24. LEAD TYPE VS REVENUE
# ==========================================================

print("\n========== LEAD TYPE VS REVENUE ==========")

lead_summary = (
    df.groupby("lead_type", dropna=False)
    .agg(
        deals=("mql_id", "count"),
        avg_revenue=("declared_monthly_revenue", "mean"),
        median_revenue=("declared_monthly_revenue", "median"),
        total_revenue=("declared_monthly_revenue", "sum")
    )
    .sort_values(
        "deals",
        ascending=False
    )
)

print(lead_summary)


# ==========================================================
# 25. REVENUE VS CATALOG SIZE
# ==========================================================

print("\n========== CATALOG SIZE VS REVENUE ==========")

print(
    df[
        [
            "declared_product_catalog_size",
            "declared_monthly_revenue"
        ]
    ].corr()
)


# ==========================================================
# 26. REVENUE VS AVERAGE STOCK
# ==========================================================

print("\n========== STOCK VS REVENUE ==========")

print(
    df[
        [
            "average_stock_numeric",
            "declared_monthly_revenue"
        ]
    ].corr()
)


# ==========================================================
# 27. NUMERICAL CORRELATION
# ==========================================================

print("\n========== NUMERICAL CORRELATION ==========")

numeric_for_corr = [
    "average_stock_numeric",
    "declared_product_catalog_size",
    "declared_monthly_revenue"
]

print(
    df[numeric_for_corr]
    .corr()
    .round(3)
)


# ==========================================================
# 28. OUTLIER CHECK - MONTHLY REVENUE
# ==========================================================

print("\n========== REVENUE OUTLIERS ==========")

Q1 = df["declared_monthly_revenue"].quantile(0.25)
Q3 = df["declared_monthly_revenue"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

revenue_outliers = df[
    (df["declared_monthly_revenue"] < lower_bound) |
    (df["declared_monthly_revenue"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

print(
    "Potential revenue outliers:",
    len(revenue_outliers)
)


# ==========================================================
# 29. OUTLIER CHECK - CATALOG SIZE
# ==========================================================

print("\n========== CATALOG SIZE OUTLIERS ==========")

Q1 = df["declared_product_catalog_size"].quantile(0.25)
Q3 = df["declared_product_catalog_size"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

catalog_outliers = df[
    (df["declared_product_catalog_size"] < lower_bound) |
    (df["declared_product_catalog_size"] > upper_bound)
]

print("Potential catalog-size outliers:",
      len(catalog_outliers))


# ==========================================================
# 30. FINAL DATA QUALITY SUMMARY
# ==========================================================

print("\n========== FINAL SUMMARY ==========")

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("Unique MQLs:",
      df["mql_id"].nunique())

print("Unique sellers:",
      df["seller_id"].nunique())

print("Unique SDRs:",
      df["sdr_id"].nunique())

print("Unique SRs:",
      df["sr_id"].nunique())

print("Unique business segments:",
      df["business_segment"].nunique())

print("Unique lead types:",
      df["lead_type"].nunique())

print("Duplicate rows:",
      df.duplicated().sum())

print("Total missing values:",
      df.isnull().sum().sum())
```
